```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/books.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Rating"].mean()
```

```python
df.sort_values(by="Rating", ascending=False).head(1)
```

```python
df["Genre"].value_counts()
```

```python
df.groupby("Genre")["Rating"].mean()
```

```python
plt.figure()
df["Genre"].value_counts().plot(kind="bar")
plt.title("Number of Books per Genre")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure()
sns.histplot(df["Rating"], bins=20)
plt.title("Distribution of Ratings")
plt.show()
```
